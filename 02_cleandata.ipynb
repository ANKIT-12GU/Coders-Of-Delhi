# Cleaning and Structuring the Data
Introduction
Your manager is impressed with your progress but points out that the data is messy. Before we can analyze it effectively, we need to clean and structure the data properly.

Your task is to:

- Handle missing values
- Remove duplicate or inconsistent data
- Standardize the data format
- Let's get started!

```python
import json
```

```python
def cleaninfo(data):
    #1 Remove user whose name in not in data,
    data['users'] = [user for user in data['users'] if user['name'].strip()]

    #2 Remove duplicate or inconsistent data
    for user in data['users']:
        user['friends'] = list(set(user['friends']))

    #3 Standardize the data format i.e must have either friends or likes pages
    data['users'] = [user for user in data['users'] if user['friends'] or user["liked_pages"]]

    print("\n Let's go for duplicate pages\n")
    # Remove duplicate pages
    Unique_pages = {}
    for page in data['pages']:
        Unique_pages[page['id']] = page
    data['pages'] = list(Unique_pages.values())
    return data

data = json.load(open("dummy.json"))
clean = cleaninfo(data)

with open("cleaned.json", "w") as f:
    json.dump(clean, f, indent=4)
print("Data has been cleaned successfully.")
```
