On social media platforms, users interact with pages by liking, following, or engaging with posts. The goal is to analyze these interactions and suggest relevant pages based on user behavior.

# What we are using here is called collaborative filtering:

- "If two people like the same thing, maybe they’ll like other things each one likes too."
- This is a basic form of a real-world recommendation engine, and our task is to implement it in pure Python. Let's Go!

# Task 2: Implement the Algorithm
We'll create a function that:

- Maps users to pages they have interacted with.
- Identifies pages liked by users with similar interests.
- Ranks recommendations based on common interactions.

```python
import json
```

```python
# 1 load the file first
def load_file(filename):
    with open(filename, "r") as file:
        return json.load(file)
```

```python
def pages_you_may_like(data, user_id):
    user_pages = {}
    # lets get the current pages he likes and store it in user_pages.
    for user in data['users']:
        user_pages[user['id']] = set(user['liked_pages'])

    if user_id not in user_pages:
        return []

    curr_liked_pages = user_pages[user_id]
    # now lets go for suggestion
    suggestion_pages = {}

    for other_user, pages in user_pages.items():
        if other_user != user_id:
            # lets find out shared pages count b/w other user and curr user like pages
            shared_pages = curr_liked_pages.intersection(pages)

        for page in pages:
            if page not in curr_liked_pages:
                suggestion_pages[page] = suggestion_pages.get(page, 0) + len(shared_pages)

    # now sort them in descreasing order accordind to score which is how many pages in comman which is shared_pages.
    sorted_suggestion = sorted(suggestion_pages.items(), key= lambda x : x[1], reverse=True)
    return [(page_id, score) for page_id, score in sorted_suggestion]     

data = load_file("massive.json")
user_id = 12
recomm_pages = pages_you_may_like(data, user_id)
print(recomm_pages)
```
