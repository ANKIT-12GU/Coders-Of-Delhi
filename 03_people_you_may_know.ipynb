# Finding "People You May Know"
Now that our data is cleaned and structured, your manager assigns you a new task: Build a 'People You May Know' feature!

In social networks, this feature helps users connect with others by suggesting friends based on mutual connections. Your job is to analyze mutual friends and recommend potential connections.

```python
import json
```

```python
def load_file(filename):
    with open(filename, "r") as file:
        return json.load(file)
```

```python
def people_you_may_know(data, user_id):
    user_friends = {}
    for user in data["users"]:
        user_friends[user['id']] = set(user['friends'])

    if user_id not in user_friends:
        return []

    current_friends = user_friends[user_id]
    for_suggestion = {}

    for friend in current_friends:
        for mutual in user_friends[friend]:
            if mutual != user_id and mutual not in current_friends:
                # If User A and User B are not friends but have mutual friends, we suggest User B to User A and vice versa.
                # Count mutual friends
                for_suggestion[mutual] = for_suggestion.get(mutual, 0) + 1

    sorted_suggestion = sorted(for_suggestion.items(), key = lambda x: x[1], reverse=True)
    return [user_id for user_id, _ in sorted_suggestion], current_friends


#data = load_file("social_data.json")
data = load_file("massive.json")
user_id = 12 
recommendations = people_you_may_know(data, user_id)
print(f"ID {user_id} has current friendship with {recommendations[1]} and\nInstagram must have to show these id {recommendations[0]} for friend suggestion.")
```
